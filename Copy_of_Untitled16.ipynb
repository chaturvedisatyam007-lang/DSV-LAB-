```python
pd.pivot_table(
    df,
    values='units_sold',
    index='branch',
    columns='product',
    aggfunc='sum',
    margins=True,
    margins_name='Total'
)
```

```python
pd.pivot_table(
    df,
    values=['discount_pct', 'customer_rating'],
    index='branch',
    columns='product',
    aggfunc='mean'
)
```

```python
print("Before Replacement:")
print(df['customer_rating'].isnull().sum())

df['customer_rating'] = df.groupby(
    ['branch', 'product']
)['customer_rating'].transform(
    lambda x: x.fillna(x.mean())
)

print("After Replacement:")
print(df['customer_rating'].isnull().sum())
```

```python
def weighted_avg(x):
    return (x['units_sold'] * x['discount_pct']).sum() / x['units_sold'].sum()

df.groupby('branch').apply(weighted_avg)
```

```python
top3 = df.groupby('branch').apply(
    lambda x: x.nlargest(3, 'units_sold')
)

top3[['branch', 'product', 'day', 'units_sold']]
```

```python
df['sales_contribution_pct'] = (
    df['units_sold'] /
    df.groupby('branch')['units_sold'].transform('sum')
) * 100

print(df[['branch', 'units_sold', 'sales_contribution_pct']])
```

```python
df['sales_category'] = pd.qcut(
    df['units_sold'],
    q=4,
    labels=[
        'Low Sales',
        'Below Average Sales',
        'Above Average Sales',
        'High Sales'
    ]
)

df.groupby('sales_category', observed=True).agg({
    'units_sold': 'sum',
    'discount_pct': 'mean',
    'customer_rating': 'mean',
    'delivery_minutes': 'mean',
    'sales_category': 'count'
})
```
