Q26. Build a consolidated class roster by concatenating section_CA.csv and section_CB.csv vertically
with ignore_index=True. Then left-merge the result with exam_marks.csv on Student_ID. Display
Student_ID, Student_Name, Section, Python_Marks and DBMS_Marks, and identify students with
missing exam marks.

```python
ca = pd.read_csv("section_CA.csv")
cb = pd.read_csv("section_CB.csv")
marks = pd.read_csv("exam_marks.csv")


roster = pd.concat(
    [ca, cb],
    ignore_index=True
)


result = pd.merge(
    roster,
    marks,
    on="Student_ID",
    how="left"
)


print(result[
    ["Student_ID", "Student_Name", "Section",
     "Python_Marks", "DBMS_Marks"]
])


missing = result[
    result["Python_Marks"].isnull() |
    result["DBMS_Marks"].isnull()
]

print("Students with missing marks:")
print(missing)
```

Q27. Combine marks_original.csv and marks_updated.csv by Student_ID using combine_first(). Merge
the resulting Marks column with students_master.csv, retaining all students. Explain, using

Q28. Create two results from marks_original.csv and marks_updated.csv after aligning on Student_ID: (a)
combine_first() and (b) combine() with np.fmax. Produce a side-by-side DataFrame showing both
results for every student and identify the IDs where they differ.

```python
original = pd.read_csv("marks_original.csv")
updated = pd.read_csv("marks_updated.csv")

original = original.set_index("Student_ID")
updated = updated.set_index("Student_ID")


result1 = original.combine_first(updated)


result2 = original.combine(updated, np.fmax)


comparison = pd.DataFrame({
    "combine_first": result1["Marks"],
    "np_fmax": result2["Marks"]
})

print(comparison)


different = comparison[
    comparison["combine_first"] != comparison["np_fmax"]
]

print("Different Student IDs:")
print(different)
```

Q29. Use repeat_tests.csv to create a pivot_table() containing both mean and maximum Marks for every
Student_ID and Subject combination. Reshape the result with stack() or unstack() to expose the
hierarchical index structure, and restore the original pivot-table layout.

```python
data = pd.read_csv("repeat_tests.csv")


result = data.pivot_table(
    index="Student_ID",
    columns="Subject",
    values="Marks",
    aggfunc=["mean", "max"]
)

print("Pivot Table:")
print(result)


stacked = result.stack(level=0)

print("After stacking:")
print(stacked)


restored = stacked.unstack(level=-1)

print("Restored table:")
print(restored)
```

Q30. Merge students_master.csv with marks_long.csv on Student_ID. Use pivot() to produce a wide table
indexed by Student_ID and Student_Name with Python, DBMS and Maths columns; then use

```python
students = pd.read_csv("students_master.csv")
marks = pd.read_csv("marks_long.csv")


merged = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="inner"
)


result = merged.pivot(
    index=["Student_ID", "Student_Name"],
    columns="Subject",
    values="Marks"
)

print("Wide format:")
print(result)

final_result = result.reset_index()

print("Final DataFrame:")
print(final_result)
```
